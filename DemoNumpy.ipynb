```python
#조건 검색
arr = np.array([10,20,30])
print(arr > 10)
```

```python
cond = arr > 10 
print( arr[cond] )
```

```python
arr = np.array([10,20,30])
cond0 = arr > 10 
cond1 = arr < 30 
print( arr[cond0 & cond1] )
```

```python
arr = np.array([10,20,30])
arr = np.where( arr > 10, 1, 0)
print(arr)
```

```python
#LG전자 주가 
lge = np.array([93000, 82400, 99100, 81000, 72300])
result = lge[ lge <= 85000 ]
print(len(result))
print(result)
```
